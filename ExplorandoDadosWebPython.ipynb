**Nível 1: Básico (URLs, Parâmetros e Cabeçalhos)**


```python
import requests
import pandas as pd
import io
from bs4 import BeautifulSoup

url = "https://jsonplaceholder.typicode.com/posts"

params = {
    "userId": 2,
    "_limit": 5
}

resposta = requests.get(url, params=params, timeout=10)

print(resposta.status_code)
print(resposta.url)
print(resposta.json())
```

```python
headers = {
    "User-Agent": "MeuProjetoPython/1.0"
}

resposta = requests.get(
    url,
    params=params,
    headers=headers,
    timeout=10
)

print(resposta.status_code)
print(resposta.url)
```

**Nível 2: Intermediário (JSON, Erros e Arquivos Binários)**

```python
ceps = [
    "24130001",
    "24130223",
    "24020125"
]

resultados = []

for cep in ceps:
    url = f"https://viacep.com.br/ws/{cep}/json/"

    resposta = requests.get(url, timeout=10)

    if resposta.status_code == 200:
        dados = resposta.json()
        resultados.append(dados)

df = pd.DataFrame(resultados)

df
```

```python
def baixar_arquivo(url, nome_arquivo):
    try:
        resposta = requests.get(url, timeout=10)
        resposta.raise_for_status()

        with open(nome_arquivo, "wb") as arquivo:
            arquivo.write(resposta.content)

        print(f"Download realizado com sucesso: {nome_arquivo}")

    except requests.exceptions.RequestException as erro:
        print(f"Erro ao realizar o download: {erro}")
```

```python
url = "https://picsum.photos/400/400"

baixar_arquivo(url, "imagem.jpg")
```

**Nível 3: Avançado (Webscraping e Ética)**

```python
url_robots = "https://books.toscrape.com/robots.txt"

resposta = requests.get(url_robots, timeout=10)

print(resposta.status_code)
print(resposta.text)
```

```python
url = "https://books.toscrape.com/"

resposta = requests.get(url, timeout=10)
resposta.raise_for_status()

soup = BeautifulSoup(resposta.text, "html.parser")
```

```python
livros = soup.select("article.product_pod")[:5]

dados_livros = []

for livro in livros:
    titulo = livro.h3.a["title"]
    preco = livro.select_one(".price_color").text

    dados_livros.append({
        "titulo": titulo,
        "preco": preco
    })

dados_livros
```

```python
df_livros = pd.DataFrame(dados_livros)

df_livros.to_csv("livros.csv", index=False)

print("Arquivo livros.csv salvo com sucesso!")
```

```python
url = "https://pt.wikipedia.org/wiki/Liga_dos_Campe%C3%B5es_da_UEFA"

pagina = requests.get(
    url,
    headers={'User-Agent': 'Mozilla/5.0'},
    timeout=20
)

print("Página baixada — status", pagina.status_code)
```
